## Day 2 — Conditions and Loops

### If / Elif / Else

```python
marks = 72

if marks >= 75:
    print("Distinction")
elif marks >= 35:
    print("Pass")
else:
    print("Fail")
```

### For Loop

```python
for number in range(1, 11):
    print(number)
```

### Multiplication Table

```python
number = 5

for i in range(1, 11):
    print(number, "x", i, "=", number * i)
```

### While Loop

```python
count = 1

while count <= 5:
    print(count)
    count += 1
```

## Day 3 — Lists, Dictionaries, and Functions

### Lists

```python
marks = [78, 65, 91, 54, 88]

print("Marks:", marks)
print("Highest:", max(marks))
print("Lowest:", min(marks))
print("Total:", sum(marks))
print("Average:", round(sum(marks) / len(marks), 2))
```

### Dictionaries

```python
student = {
    "name": "Rahul",
    "age": 20,
    "marks": 85
}

print("Name:", student["name"])
print("Age:", student["age"])
print("Marks:", student["marks"])
```

### Functions

```python
def calculate_percentage(obtained, total_marks):
    return (obtained / total_marks) * 100

result = calculate_percentage(425, 500)

print("Percentage:", round(result, 2), "%")
```

## Day 4 — File Handling and Data Science Introduction

### Write to a File

```python
with open("students.txt", "w") as file:
    file.write("Rahul,85\n")
    file.write("Anita,92\n")
    file.write("Kiran,76\n")

print("students.txt created successfully.")
```

### Read from a File

```python
with open("students.txt", "r") as file:
    data = file.read()

print(data)
```

```python
students = {
    "Rahul": 78,
    "Anita": 92,
    "Kiran": 54,
    "Priya": 88,
    "Arjun": 31,
    "Meena": 67
}

scores = list(students.values())
class_average = sum(scores) / len(scores)

highest_student = max(students, key=students.get)
lowest_student = min(students, key=students.get)

passed = []
failed = []
above_average = []

for student_name, score in students.items():
    if score >= 35:
        passed.append(student_name)
    else:
        failed.append(student_name)

    if score > class_average:
        above_average.append(student_name)

print("Total Students:", len(students))
print("Class Average:", round(class_average, 2))
print("Highest:", highest_student, "-", students[highest_student])
print("Lowest:", lowest_student, "-", students[lowest_student])
print("Passed:", len(passed), passed)
print("Failed:", len(failed), failed)
print("Above Average:", above_average)
```
