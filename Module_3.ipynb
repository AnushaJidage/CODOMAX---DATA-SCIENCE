# Day 9 — Setup and Visualization Fundamentals



```python
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns

print("NumPy version:", np.__version__)
print("Pandas version:", pd.__version__)
print("Seaborn version:", sns.__version__)
```

```python
data = {
    "Student": ["Rahul", "Anita", "Kiran", "Priya", "Arjun",
                "Meena", "Sara", "Vijay", "Neha", "Rohan"],
    "Maths": [78, 92, 65, 88, 52, 67, 84, 73, 95, 61],
    "Science": [74, 89, 68, 91, 58, 72, 81, 70, 93, 64],
    "Study_Hours": [4, 6, 3, 5.5, 2, 3.5, 5, 4, 6.5, 2.5],
    "Attendance": [88, 96, 82, 94, 71, 85, 92, 87, 98, 76],
    "City": ["Bengaluru", "Mysuru", "Hubballi", "Kalaburagi",
             "Bengaluru", "Mysuru", "Kalaburagi", "Hubballi",
             "Bengaluru", "Kalaburagi"]
}

df = pd.DataFrame(data)
df
```

# Day 9 — Bar Chart

```python
plt.figure(figsize=(10, 5))

plt.bar(df["Student"], df["Maths"])

plt.title("Maths Scores by Student")
plt.xlabel("Student")
plt.ylabel("Maths Score")
plt.xticks(rotation=45)
plt.tight_layout()

plt.show()
```

## Seaborn Bar Chart


```python
plt.figure(figsize=(10, 5))

sns.barplot(data=df, x="Student", y="Maths")

plt.title("Maths Scores by Student — Seaborn")
plt.xlabel("Student")
plt.ylabel("Maths Score")
plt.xticks(rotation=45)
plt.tight_layout()

plt.show()
```

# Day 10 — Line Chart



```python
tests = ["Test 1", "Test 2", "Test 3", "Test 4", "Test 5"]
scores = [62, 68, 74, 81, 88]

plt.figure(figsize=(8, 5))

plt.plot(tests, scores, marker="o")

plt.title("Student Performance Across Tests")
plt.xlabel("Test")
plt.ylabel("Score")
plt.grid(True)

plt.show()
```

# Day 10 — Pie Chart


```python
city_counts = df["City"].value_counts()

plt.figure(figsize=(7, 7))

plt.pie(
    city_counts.values,
    labels=city_counts.index,
    autopct="%1.1f%%",
    startangle=90
)

plt.title("Student Distribution by City")

plt.show()
```

# Day 11 — Histogram



```python
plt.figure(figsize=(8, 5))

plt.hist(df["Maths"], bins=5, edgecolor="black")

plt.title("Distribution of Maths Scores")
plt.xlabel("Score Range")
plt.ylabel("Number of Students")

plt.show()
```

## Histogram with Seaborn


```python
plt.figure(figsize=(8, 5))

sns.histplot(data=df, x="Maths", bins=5, kde=True)

plt.title("Distribution of Maths Scores — Seaborn")
plt.xlabel("Maths Score")
plt.ylabel("Frequency")

plt.show()
```

# Day 11 — Scatter Plot


```python
plt.figure(figsize=(8, 5))

plt.scatter(df["Study_Hours"], df["Maths"])

plt.title("Study Hours vs Maths Score")
plt.xlabel("Study Hours")
plt.ylabel("Maths Score")
plt.grid(True)

plt.show()
```

## Scatter Plot with Seaborn


```python
plt.figure(figsize=(8, 5))

sns.scatterplot(
    data=df,
    x="Study_Hours",
    y="Maths",
    s=100
)

plt.title("Study Hours vs Maths Score — Seaborn")

plt.show()
```

```python
correlation = df["Study_Hours"].corr(df["Maths"])

print("Correlation between Study Hours and Maths Score:",
      round(correlation, 3))
```

# Day 12 — Multiple Variable Visualization



```python
plt.figure(figsize=(10, 5))

x = np.arange(len(df))
width = 0.35

plt.bar(x - width/2, df["Maths"], width, label="Maths")
plt.bar(x + width/2, df["Science"], width, label="Science")

plt.xticks(x, df["Student"], rotation=45)
plt.xlabel("Student")
plt.ylabel("Score")
plt.title("Maths vs Science Performance")
plt.legend()
plt.tight_layout()

plt.show()
```

# Day 12 — Attendance vs Performance



```python
plt.figure(figsize=(8, 5))

sns.scatterplot(
    data=df,
    x="Attendance",
    y="Maths",
    size="Study_Hours",
    sizes=(50, 250)
)

plt.title("Attendance vs Maths Score")
plt.xlabel("Attendance (%)")
plt.ylabel("Maths Score")

plt.show()
```

## Correlation Matrix



```python
numeric_df = df[["Maths", "Science", "Study_Hours", "Attendance"]]

correlation_matrix = numeric_df.corr()

plt.figure(figsize=(7, 5))

sns.heatmap(
    correlation_matrix,
    annot=True,
    fmt=".2f"
)

plt.title("Correlation Matrix")

plt.show()
```

```python
top_student = df.loc[df["Maths"].idxmax()]

print("VISUALIZATION PROJECT — KEY FINDINGS")
print("-" * 50)

print("Top Maths Student:",
      top_student["Student"],
      "-", top_student["Maths"])

print("Average Maths Score:",
      round(df["Maths"].mean(), 2))

print("Average Science Score:",
      round(df["Science"].mean(), 2))

print("Study Hours vs Maths Correlation:",
      round(df["Study_Hours"].corr(df["Maths"]), 3))

print("Attendance vs Maths Correlation:",
      round(df["Attendance"].corr(df["Maths"]), 3))
```
