```python
practice_data = {
    "Student_ID": [101,102,103,104,105,106,106,107,108],
    "Name": ["Rahul","Anita","Kiran","Priya","Arjun","Meena","Meena","Sara","Vijay"],
    "Marks": [78,92,np.nan,88,31,67,67,84,np.nan],
    "City": ["Bengaluru","Mysuru","Hubballi","Kalaburagi",None,"Mysuru","Mysuru","Kalaburagi","Hubballi"]
}
df = pd.DataFrame(practice_data)
df.to_csv("student_performance.csv", index=False)
df = pd.read_csv("student_performance.csv")
df
```

## Day 7 — Data Cleaning

```python
print("Missing values:")
print(df.isnull().sum())
print("\nDuplicate rows:", df.duplicated().sum())
```

```python
clean_df = df.drop_duplicates().copy()
print("Rows before:", len(df))
print("Rows after duplicate removal:", len(clean_df))
```

```python
median_marks = clean_df["Marks"].median()
clean_df["Marks"] = clean_df["Marks"].fillna(median_marks)
clean_df["City"] = clean_df["City"].fillna("Unknown")

print("Median used for missing Marks:", median_marks)
print("\nMissing values after cleaning:")
print(clean_df.isnull().sum())
```

```python
clean_df["Marks"] = pd.to_numeric(clean_df["Marks"], errors="coerce")
clean_df = clean_df.sort_values("Student_ID").reset_index(drop=True)
clean_df
```

## Day 8 — Basic Data Analysis

```python
print("Average:", round(clean_df["Marks"].mean(), 2))
print("Median:", clean_df["Marks"].median())
print("Highest:", clean_df["Marks"].max())
print("Lowest:", clean_df["Marks"].min())
```

```python
clean_df["Status"] = np.where(clean_df["Marks"] >= 35, "Pass", "Fail")
display(clean_df[["Name","Marks","Status"]])
print(clean_df["Status"].value_counts())
```

```python
high_scorers = clean_df[clean_df["Marks"] >= 75].sort_values("Marks", ascending=False)
high_scorers
```

```python
city_analysis = clean_df.groupby("City", as_index=False).agg(
    Students=("Student_ID","count"),
    Average_Marks=("Marks","mean"),
    Highest_Marks=("Marks","max")
)
city_analysis["Average_Marks"] = city_analysis["Average_Marks"].round(2)
city_analysis
```

## Key Findings

```python
top_student = clean_df.loc[clean_df["Marks"].idxmax()]
lowest_student = clean_df.loc[clean_df["Marks"].idxmin()]
total = len(clean_df)
passed = (clean_df["Status"] == "Pass").sum()
failed = (clean_df["Status"] == "Fail").sum()

print("Total students:", total)
print("Average marks:", round(clean_df["Marks"].mean(), 2))
print("Pass count:", passed)
print("Fail count:", failed)
print("Pass rate:", round(passed / total * 100, 2), "%")
print("Top performer:", top_student["Name"], "-", top_student["Marks"])
print("Lowest performer:", lowest_student["Name"], "-", lowest_student["Marks"])
```
